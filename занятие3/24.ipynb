**Задача 1**

Текстовый файл `24_1.txt` состоит из заглавных букв латинского алфавита Q,R,W   и цифр 1, 2, 4. Определите в прилагаемом файле максимальное количество идущих подряд символов, среди которых ни одна буква не стоит рядом с буквой, а цифра – с цифрой. Для выполнения этого задания следует написать программу.

```python
with open('24_1.txt', encoding='utf-8') as f:
    s = f.read().strip()

def is_letter(ch):
    return ch in 'QRW'

def is_digit(ch):
    return ch in '124'

max_len = 0
cur_len = 1

for i in range(1, len(s)):
    # Соседние символы должны быть разных типов: буква + цифра
    if (is_letter(s[i]) and is_digit(s[i - 1])) or \
       (is_digit(s[i]) and is_letter(s[i - 1])):
        cur_len += 1
    else:
        cur_len = 1

    if cur_len > max_len:
        max_len = cur_len


print(max_len)
```

**Задача 2**

Текстовый файл `24_2.txt` состоит не более чем из 1 200 000 символов A  , B  , C   и D  .

Определите максимальное количество идущих подряд символов, среди которых нет символов A   и В, стоящих рядом.

Для выполнения этого задания следует написать программу.



```python
with open('24_2.txt', encoding='utf-8') as f:
    s = f.read().strip()

max_len = 1
cur_len = 1

for i in range(1, len(s)):
    # Запрещена только пара A-B или B-A
    if (s[i - 1] == 'A' and s[i] == 'B') or (s[i - 1] == 'B' and s[i] == 'A'):
        cur_len = 1
    else:
        cur_len += 1

    if cur_len > max_len:
        max_len = cur_len

if len(s) == 0:
    max_len = 0

print(max_len)
```

**Задача 3**

Текстовый файл `24_3.txt` состоит из символов, обозначающих прописные буквы латинского алфавита.

Определите максимальное количество идущих подряд символов, в которых никакие две буквы из набора букв  Q  , R   и S   (с учетом повторений) не записаны подряд.

Для выполнения этого задания следует написать программу.

```python
with open('24_3.txt', encoding='utf-8') as f:
    s = f.read().strip()

letters = set('QRS')

max_len = 0
cur_len = 1

if s:
    max_len = 1
    for i in range(1, len(s)):
        # Если два соседних символа оба принадлежат Q, R или S,
        # то они не могут находиться в одной допустимой последовательности
        if s[i - 1] in letters and s[i] in letters:
            cur_len = 1
        else:
            cur_len += 1

        if cur_len > max_len:
            max_len = cur_len

print(max_len)
```

**Задача 4**

Текстовый файл `24_1.txt` состоит не более чем из 1 200 000 символов X  , Y  , и Z  .

Определите максимальное количество идущих подряд символов, среди которых нет подстроки XZZY.

Для выполнения этого задания следует написать программу.

```python
with open('24_1.txt', encoding='utf-8') as f:
    s = f.read().strip()

pattern = 'XZZY'
start = 0
max_len = 0

for i in range(len(s)):
    # Если заканчивается запрещённая подстрока XZZY
    if i >= 3 and s[i - 3:i + 1] == pattern:
        # Следующая допустимая последовательность может начаться
        # только после символа X, то есть с позиции i - 2
        start = i - 2

    max_len = max(max_len, i - start + 1)

print(max_len)
```

**Задача 6**


Текстовый файл `24_6.txt` состоит из символов A  , C  , D  ,F   и O  .

Определите максимальное количество идущих подряд пар символов вида

гласная + согласная

в прилагаемом файле.

Для выполнения этого задания следует написать программу.

```python
with open('24_6.txt', encoding='utf-8') as f:
    s = f.read().strip()

vowels = set('AO') 

ans = 0
cur = 0

for i in range(len(s) - 1):
    # Проверяем пару s[i], s[i+1] вида "гласная + согласная"
    if s[i] in vowels and s[i + 1] not in vowels:
        # Может ли эта пара продолжить предыдущую пару?
        # Предыдущая пара должна заканчиваться на позиции i-1,
        # то есть занимать позиции i-2, i-1.
        if i >= 2 and s[i - 2] in vowels and s[i - 1] not in vowels:
            cur += 1
        else:
            cur = 1

        ans = max(ans, cur)

print(ans)
```

**Задача 8**

Текстовый файл `24_8.txt` состоит из цифр 0, 2, 3, 4, 5 и знаков арифметических операций «− » и «*» (вычитание и умножение). Определите максимальное количество символов в непрерывной последовательности, которая является корректным арифметическим выражением с целыми неотрицательными числами. В этом выражении никакие два знака арифметических операций не стоят рядом, в записи чисел отсутствуют незначащие (ведущие) нули и число 0 не имеет знака.

В ответе укажите количество символов.



```python

with open('8.txt', encoding='utf-8') as f:
    s = f.read().strip()

DIGITS = set('02345')
NEG = -10**18

zero = num = after = NEG

ans = 0

for ch in s:
    n_zero = n_num = n_after = NEG

    if ch in DIGITS:
        if ch == '0':
            n_zero = max(n_zero, 1)
        else:
            n_num = max(n_num, 1)

        if after != NEG:
            if ch == '0':
                n_zero = max(n_zero, after + 1)
            else:
                n_num = max(n_num, after + 1)

        if num != NEG:
            n_num = max(n_num, num + 1)

    else:  
        best = max(zero, num)
        if best != NEG:
            n_after = best + 1

    zero, num, after = n_zero, n_num, n_after

    ans = max(ans, zero, num)

print(ans)
```

Назовём маской числа последовательность цифр, в которой также могут встречаться следующие символы:

– символ «?» означает ровно одну произвольную цифру;

– символ «\*» означает любую последовательность цифр произвольной длины; в том числе «*» может задавать и пустую последовательность.

Например, маске 123*4?5 соответствуют числа 123405 и 12300405.

Среди натуральных чисел, не превышающих 10^10  , найдите все числа, соответствующие маске 12345?7?8, делящиеся на число 23 без остатка.

В ответ запишите пары найденных чисел (найденное число и частное от деления на 23) в порядке возрастания самих чисел.

```python
from fnmatch import fnmatch  

for x in range(123450708, 123459798 + 1):
    if fnmatch(str(x), "12345?7?8") and x % 23 == 0:
        print(x, x // 23)
```

— символ «*» означает любую последовательность цифр произвольной длины; в том числе «*» может задавать и пустую последовательность.

Сколько существует натуральных чисел, соответствующих следующей маске
123?123?123   и разность между суммой цифр, стоящих на нечетных местах, и суммой цифр, стоящих на четных местах делилась на 11  . В ответ запишите количество таких чисел.

```python
from fnmatch import fnmatch
def f(x):
    l = str(x)
    v1 = 0
    v2=0
    for i in range(1,len(l)+1):
        if i%2:
            v1 += int(l[-i])
        else:
            v2 += int(l[-i])

    return v1 == v2


c= 0
for x in range(12301230128,12391239123+1,11):
    if fnmatch(str(x), "123?123?123") and f(x) and x%11 == 0:
        c+=1

print(c)
```
